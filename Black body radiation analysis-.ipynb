```python
# All variables in SI units
h = 6.63e-34# Planck constant
c=3.00e8     # Speed of light
kB=1.38e-23  # Boltzmann constant
T = 300

import numpy as np
import matplotlib.pyplot as plt

def rho_class(nu):
    return 8*np.pi*kB*nu**2*T / c**3

def rho_quant(nu):
    return (8*np.pi*h*nu**3) / (c**3 * (np.exp(h*nu/(kB*T)) - 1))

nu = np.linspace(0.1, 1e14, 500) # nu axis from 0 and 10^14 Hz in 500 even steps

plt.figure(figsize=(10,6))
plt.plot(nu, rho_quant(nu), label="Quantum")
plt.plot(nu, rho_class(nu), label="Classical")

plt.xlabel("frequency (Hz)")
plt.ylabel("energy density")

plt.ylim(0, np.max(rho_quant(nu))*1.1)
plt.xlim(0, 1e14)

plt.legend()
plt.grid()
plt.show()
# All variables in SI units
h = 6.63e-34   # Planck constant
c = 3.00e8     # Speed of light 
kB = 1.38e-23  # Boltzmann constant
T=300

# Visible spectrum bounds converted to frequency

lambda_blue = 380e-9   # 380 nm
lambda_red  = 750e-9   # 750 nm

nu_Blue = round(c / lambda_blue, -13)   # nearest 10 THz
nu_Red  = round(c / lambda_red,  -13)

print("nu_Blue =", nu_Blue)
print("nu_Red  =", nu_Red)

# Temperatures

T_RedHot = 1000       # typical red-hot metal
T_WhiteHot = 2000     # typical white-hot metal
T_Sun = 5760          # visible surface of the Sun

# Frequency axis (template)

nu = np.linspace(nu_Red, nu_Blue, 500) # Frequency axis between nu_Red and nu_Blue in 500 even steps

# rho_quant(nu, T) function (correct Planck formula)

def rho_quant(nu, T):
    x = h*nu/(kB*T)
    # For large x, exp(x) overflows, but the whole fraction tends to 0
    with np.errstate(over='ignore'):
        return (8*np.pi*h*nu**3) / (c**3 * (np.exp(x) - 1))


fig = plt.figure()
ax = fig.add_axes([0, 0, 1, 1])

ax.plot(nu, rho_quant(nu, T), label="300 K (room temp)")
ax.plot(nu, rho_quant(nu, T_RedHot), label="Red hot (~1200 K)")
ax.plot(nu, rho_quant(nu, T_WhiteHot), label="White hot (~2500 K)")
ax.plot(nu, rho_quant(nu, T_Sun), label="Sun (~5800 K)")

ax.set_xlabel("frequency (Hz)")
ax.set_ylabel("energy density ρ(ν)")
ax.legend()

plt.show()

# Logarithmic vertical axis version

fig = plt.figure()
ax = fig.add_axes([0, 0, 1, 1])
ax.set_yscale('log')

ax.plot(nu, rho_quant(nu, T), label="300 K")
ax.plot(nu, rho_quant(nu, T_RedHot), label="Red hot")
ax.plot(nu, rho_quant(nu, T_WhiteHot), label="White hot")
ax.plot(nu, rho_quant(nu, T_Sun), label="Sun")

ax.set_xlabel("frequency (Hz)")
ax.set_ylabel("energy density ρ(ν) [log scale]")
ax.legend()

plt.show()
```
