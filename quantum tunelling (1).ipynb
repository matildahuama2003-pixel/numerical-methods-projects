```python

import numpy as np
import matplotlib.pyplot as plt
m = 9.109e-31     # Electron mass
hb = 1.05e-34     # Reduced Planck constant
eV = 1.6e-19      # One electron volt in Joules
nm = 1e-9         # One nanometer in meters

b = 1.4*nm        # width of the barrier
v0 = 1*eV         # The barrier potential V_0
E = 0.9*eV        # Energy of the particle

# Setting up the position axis, as an array, and the three relevant regions
xAxis = np.linspace(-5*nm, b + 5*nm, 1000)
xL = xAxis[xAxis < 0]
xB = xAxis[(xAxis >= 0) & (xAxis <= b)]
xR = xAxis[xAxis > b]

# Wave numbers
k = np.sqrt(2*m*E) / hb
alpha = np.sqrt(2*m*(v0 - E)) / hb

# Continuity conditions
A = 1
B = np.exp(-alpha*b) * np.exp(-1j*k*b)

def Psi_L(x):
    return np.exp(1j*k*x)

def Psi_b(x):
    return A * np.exp(-alpha*x)

def Psi_R(x):
    return B * np.exp(1j*k*x)

# Plotting
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
    
ax.plot(xL/nm, np.real(Psi_L(xL)), label="Region L (x < 0)")
ax.plot(xB/nm, np.real(Psi_b(xB)), label="Barrier (0 < x < b)")
ax.plot(xR/nm, np.real(Psi_R(xR)), label="Region R (x > b)")

ax.set_xlabel("x (nm)")
ax.set_ylabel("Re[ψ(x)]")
ax.set_title("Quantum Tunnelling Wavefunction")
ax.legend()

plt.show()
```
